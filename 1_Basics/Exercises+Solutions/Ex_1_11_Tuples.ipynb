## 🟨 Add Job Role to Tuple (1.11.2)

```python
# Code Needed
job_roles = ('Data Scientist', 'Data Analyst', 'Machine Learning Engineer')
# by convering tuple to list, editting list and converting it back to tuple
job_roles_list = list(job_roles)
job_roles_list.append("AI Specialist")
extended_job_roles = tuple(job_roles_list)
extended_job_roles
```

```python
# or easier way
job_roles = ('Data Scientist', 'Data Analyst', 'Machine Learning Engineer')
additional = ("AI Specialist",) #In Python, it is the trailing comma , that creates a tuple, not the parentheses!
                              # To make a tuple with only one item, you must add a comma after the item inside the parentheses
extended_job_roles = job_roles + additional
extended_job_roles
```

## 🟨 Unpack Job Postings Tuple (1.11.4)

```python
# Code Needed
job_postings = (120, 80, 50)
data_scientist_postings, data_analyst_postings, ml_engineer_postings = job_postings
#We use multiple assignment to unpack the job_postings tuple into three variables: data_scientist_postings, data_analyst_postings, and ml_engineer_postings.
print(data_scientist_postings)
print(data_analyst_postings)
print(ml_engineer_postings)
```

## 🟥 Extract Nested Value from Tuple (1.11.5)

```python
# Code Needed
# Extract and print the number of postings for 'Machine Learning Engineer' from the nested tuple job_details.
job_details = (('Data Scientist', 120), ('Data Analyst', 80), ('Machine Learning Engineer', 50)) #inside the tuple job_details we have tuples
variable = job_details[2] #variable is  tuple extracted from index 2
print(variable[1]) # by variable[1] I access the item in tuple
```

```python
# or you have been done like this
job_details = (('Data Scientist', 120), ('Data Analyst', 80), ('Machine Learning Engineer', 50))
print(job_details[2][1])
# We use indexing to access the third item in the job_details tuple, which is the tuple for 'Machine Learning Engineer'.
# We use indexing again to access the second item within that nested tuple, which is the number of postings.
```
