## 🟩 Total Job Postings (1.13.2)

```python
# Code Needed
job_postings = [5, 10, 15, 20, 25]
index = 0
total_job_postings = 0
while index <=4:
  total_job_postings += job_postings[index]
  index += 1
print(total_job_postings)
```

## 🟨 Print Job Titles with Index (1.13.3)

```python
# Code Needed
job_titles = ['Data Scientist', 'Data Analyst', 'Machine Learning Engineer']
# Instead of simply iterating over the list with `x in list` to get the job titles,
# we can use `enumerate` to get both the index and the job titles
# syntax is `enumerate(iterable, start = 0)`
for index, job_title  in enumerate(job_titles, start=0): # we start from zero because in task it was asked to print indexes and indexes start with 0
  print(f"{index} : {job_title}")
```

## 🟨 Find First Engineer (1.13.4)

```python
# Code Needed
job_titles = ['Data Scientist', 'Data Analyst', 'Machine Learning Engineer', 'Software Engineer']
index = 0
while index < len(job_titles):
  if "Engineer" in job_titles[index]:
    print(job_titles[index])
    break
  else:
    index += 1
```

## 🟥 Print Job Titles Containing 'Data' (1.13.5)

```python
# Code Needed
#  |  items(self, /)
#  |      Return a set-like object providing a view on the dict's items.

positions_skills = {
    'Data Scientist': ['Python', 'R'],
    'Data Analyst': ['SQL', 'Excel'],
    'Machine Learning Engineer': ['Python', 'TensorFlow']
}

for position, skills in positions_skills.items():
  for skill in skills:
    print(f"The role of {position} requires proficiency in {skill}")
```
