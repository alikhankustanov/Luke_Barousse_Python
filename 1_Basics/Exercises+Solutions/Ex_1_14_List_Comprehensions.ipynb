## 🟩 Convert Salaries to EUR (1.14.1)

```python
# Code Needed
salaries_usd = [100000, 120000, 80000, 90000]
salaries_euro = [0.85*x for x in salaries_usd]
salaries_euro
```

## 🟨 Create List of Data Jobs (1.14.2)

```python
# Code Needed
job_experience = {'Data Scientist': 3, 'Data Analyst': 1, 'Machine Learning Specialist': 4, 'Data Engineer': 2}
filter_job = [[key, value] for key, value in job_experience.items() if value > 2] #.items() function to acces both key and value
filter_job
# in this case extracted both the key and the value
# [key, value] is a list because i extract both key and value. if i wasnt extracting both then I wouldnt need to put the []
```

## 🟥 Extract High Salary Jobs (1.14.3)

```python
# Code Needed
job_salary_dicts = [{'job_title': 'Data Scientist', 'salary': 100000}, {'job_title': 'Data Analyst', 'salary': 120000}, {'job_title': 'Data Engineer', 'salary': 80000}, {'job_title': 'Machine Learning Engineer', 'salary': 90000}]
# so we have a list containing dictionaries
high_salary_jobs =[x["job_title"] for x in job_salary_dicts if x["salary"] > 90_000]
high_salary_jobs
#x[job_title] extract the value of dictionary with key "job_title"

#The Loop (for x in job_salary_dicts):
# This tells Python to iterate through your list. In each iteration, the variable x represents one entire dictionary from the list
#  (e.g., {'job_title': 'Data Scientist', 'salary': 100000}).

# The Condition (if x["salary"] > 90_000):
# For every dictionary (x) it looks at, Python checks the value associated with the "salary" key. If that number is strictly greater than 90,000, it passes the filter.

# The Output Expression (x["job_title"]):
# If the dictionary passes the condition above, this part tells Python what to actually keep. Instead of keeping the whole dictionary x,
# it extracts just the string value tied to the "job_title" key and places it into the new list.
```
