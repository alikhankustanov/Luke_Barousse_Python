## 🟩 Job Title Contains (1.16.1)

```python
# Code Needed
job_title = 'Data Scientist'
keyword = 'Data'
def job_title_contains(job_title, keyword):
  return keyword in job_title
# The in operator is a built-in Python feature used for membership testing. When used with strings,
# it checks if the string on the left (keyword) is a substring of the string on the right (job_title).
# This operation evaluates to a Boolean value, meaning it will result in either True or False.
```

```python
job_title_contains(job_title, keyword)
```

## 🟨 Salary Statistics (1.16.3)

```python
# Code Needed
# Create a function salary_statistics that takes a list of salaries and returns a dictionary with the minimum, maximum, and average salary
salaries = [95000, 120000, 105000, 90000, 130000]
def salary_statistics(salaries):
  sum(salaries)/len(salaries)
  min(salaries)
  max(salaries)
  dictionary = {"minimum": min(salaries), "maximum": max(salaries), "average": sum(salaries)/len(salaries)}
  return(dictionary)
```

```python
salary_statistics(salaries)
```

```python
# shorter way to do
def salary_statistics(salaries):
    return {
        'minimum': min(salaries),
        'maximum': max(salaries),
        'average': sum(salaries) / len(salaries)
    }

salary_statistics(salaries)
```

## 🟥 Job Posting Summary (1.16.4)

```python
# Code Needed
# Create a function job_posting_summary that takes a list of job postings, where each posting is a dictionary with keys 'title', 'location', and 'salary',
# and returns a summary dictionary with the total number of postings, the average salary, and a list of unique locations.
job_postings = [
    {'title': 'Data Scientist', 'location': 'New York', 'salary': 95000},
    {'title': 'Data Analyst', 'location': 'San Francisco', 'salary': 85000},
    {'title': 'Machine Learning Engineer', 'location': 'New York', 'salary': 115000}
]
def job_posting_summary(job_postings):
  total_postings = len(job_postings) # inside a list we have 3 dictionaries
  total_salary = sum(posting["salary"] for posting in job_postings) # for posting(its a dictionary) in job_postings. posting["salary"] is to extract value of salary from dictionary
  average_salary = total_salary / total_postings
  unique_locations = list(set(posting["location"] for posting in job_postings))
  # unique locations is a list of set. you make set from posting["location"] for posting in job_postings
  return {
      "total number of postings": total_postings,
      "average salary": average_salary,
      "unique locations": unique_locations
  }
```

---

```python
job_posting_summary(job_postings)
```
