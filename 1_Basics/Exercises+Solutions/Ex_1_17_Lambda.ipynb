## 🟩 Calculate Average Salary (1.17.1)

```python
# Code Needed
# He said no need to use lambda if dont understand so I use user defined fucntion
salaries = [95000, 120000, 105000, 90000, 130000]
def average_salary(salaries):
  average_salary = sum(salaries)/len(salaries)
  return average_salary
```

```python
average_salary(salaries)
```

## 🟨 Filter Job Titles (1.17.2)

```python
# Code Needed
# He said no need to use lambda if dont understand so I use user defined fucntion
job_titles = ['Data Scientist', 'Data Engineer', 'Machine Learning Engineer', 'Data Analyst']
def data_filtering(job_titles):
  a = list((job for job in job_titles if "Data" in job))
  return a
```

```python
data_filtering(job_titles)
```

## 🟥 Filter Remote Python Jobs (1.17.3)

```python
# Code Needed
# He said no need to use lambda if dont understand so I use user defined fucntion
job_postings = [
    {'title': 'Data Scientist', 'skills': ['Python', 'SQL'], 'remote': True},
    {'title': 'Data Analyst', 'skills': ['Excel', 'SQL'], 'remote': False},
    {'title': 'Machine Learning Engineer', 'skills': ['Python', 'TensorFlow'], 'remote': True},
    {'title': 'Software Developer', 'skills': ['Java', 'C++'], 'remote': True}
]
def python_remote(job_postings):
  a = list((job["title"] for job in job_postings if "Python" in job["skills"] and job["remote"] == True))
  return a
```

```python
python_remote(job_postings)
```
