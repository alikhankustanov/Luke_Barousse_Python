## 🟥 Fetch Data from URL (1.20.4)

```python
# Use the requests library to extract data from the URL 'https://pokeapi.co/api/v2/pokemon/pikachu' and print the JSON response.
# Code Needed
url = 'https://pokeapi.co/api/v2/pokemon/pikachu'

import requests
response = requests.get(url)
data = response.json()
print(data)

# Import the requests library.
# Use the requests.get method to fetch or extract data from the URL.
# Use the json method of the response object to get the JSON response.
# Print the JSON response.
```
