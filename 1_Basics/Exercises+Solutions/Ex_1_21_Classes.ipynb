## 🟩 Create JobPosting Class (1.21.1)

```python
# Create the JobPostings class in this code cell
```

```python
# Create a class named JobPosting with attributes title, company, and location.
# Create an instance of an object (called job) of this class with the title 'Data Scientist', company 'Tech Innovations', and location 'New York'.
# Print the attributes of the object.

class JobPosting:
  def __init__(self, title, company, location):
    self.title = title
    self.company = company
    self.location = location
```

```python
job = JobPosting('Data Scientist', 'Tech Innovations', 'New York')
```

```python
print(job.title)
print(job.company)
print(job.location)
```

The Blueprint
The class JobPosting: line establishes the basic template for any future job postings you want to track in your system.

The Initializer
The __init__ method acts as the catching mechanism. It takes the three mandatory arguments (title, company, location) alongside the automatic self reference, and permanently saves them to the specific object using the self.attribute_name = value syntax.

Instantiation
Writing job = JobPosting(...) creates a live, concrete object from your blueprint and assigns it to the variable job. Because you did not define any default values in your __init__ method (like you did with bonus_rate=0.1 previously), providing exactly those three string arguments here is strictly required.

Attribute Access
The final print() statements demonstrate dot notation (job.title). This simply reaches inside the job object to retrieve the specific data stored in its memory during the initialization step.

## 🟨 Convert to Dictionary (1.21.2)

```python
# Create the JobPostings class in this code cell
```

```python
# Create a class named JobPosting with a method to_dict that converts the job posting details to a dictionary.
# Instantiate an object (called job) of this class with the title 'Data Scientist', company 'Tech Innovations', location 'New York', and salary as 120000.
# Convert its details to a dictionary using the to_dict method.
# Print the dictionary.
class JobPosting:

  def __init__(self, title, company, location, salary):
    self.title = title
    self.company = company
    self.location = location
    self.salary = salary

  def to_dict(self): #create a method
    return {
            'title': self.title,
            'company': self.company,
            'location': self.location,
            'salary': self.salary
        }
```

```python
job = JobPosting('Data Scientist', 'Tech Innovations', 'New York', 120000)
```

```python
print(job.to_dict())
```

The Initializer (def __init__(self, title, company, location, salary):)
This is the setup phase that runs automatically the moment a new object is created.

- self acts as the blank canvas for the specific object being built.

- The parameters (title, company, etc.) act as catchers for the raw data you pass in.

- Instructions like self.title = title take the temporary data on the right and permanently glue it to the object's memory on the left as an attribute, making it available for the rest of the object's lifespan.

The Custom Method (def to_dict(self):)
While __init__ sets up data, this method defines a specific action the object can perform. By requiring self as a parameter, the method is granted full access to read all the attributes you saved earlier. It constructs a standard Python dictionary ({}), creating string keys (like 'title') and assigning them the values pulled directly from the object's memory (self.title).

Instantiation (job = JobPosting(...))
This line transforms the abstract blueprint into a concrete reality. You are creating a unique instance of a JobPosting and injecting it with four specific pieces of data. Python runs the __init__ method silently in the background, locking "Data Scientist", "Tech Innovations", "New York", and 120000 into the newly created job object.

Method Execution (print(job.to_dict()))
Here, you command the specific job object to execute its to_dict action. The object reaches into its own saved attributes, builds the formatted dictionary, and hands it back. The outer print() function simply takes that finished dictionary and displays it in your console.

## 🟥 Compare Job Salaries (1.21.3)

```python
# Create the JobPostings class in this code cell
```

```python
# Create a class named JobPosting with a method compare_salary that compares the salary of two job postings and returns the title of the job with the higher salary.
# Instantiate two objects called job1 and job2 of this class and compare their salaries using the compare_salary method.
# The two objects are defined below in the code needed.
class JobPosting:

  def __init__(self, title, company, location, salary):
    self.title = title
    self.company = company
    self.location = location
    self.salary = salary

  def compare_salary(self, other_job):
    if self.salary > other_job.salary:
      return self.title
    else:
      return other_job.title
```

```python
job1 = JobPosting('Data Scientist', 'Tech Innovations', 'New York', 120000)
job2 = JobPosting('Data Analyst', 'Data Driven Co', 'San Francisco', 100000)
```

```python
print(job1.compare_salary(job2))
```

other_job should not be initialized in the __init__ dunder method because it is not a permanent characteristic of a job posting.

When you initialize something using self (like self.title = title), you are defining the core, permanent data that belongs exclusively to that specific object. A job posting inherently possesses a title, a company, a location, and a salary. It does not inherently own another job posting.

Instead of being a permanent attribute, other_job acts as a temporary visitor for a specific action.

- Attributes (self.salary): This is permanent data the object remembers for its entire lifespan.

- Method Arguments (other_job): This is temporary data passed into the compare_salary method just to perform a quick calculation. Once the comparison is finished, the method stops running and forgets about other_job.
