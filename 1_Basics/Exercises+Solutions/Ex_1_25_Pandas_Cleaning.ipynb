```python
!pip install -U datasets
import pandas as pd
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
```

```python
df
```

---

## 🟩 Sort by Salary (1.25.1)

```python
# Sort the DataFrame by salary_year_avg in descending order and display the first 10 rows.
df.sort_values("salary_year_avg", axis = 0, ascending=False, inplace=True)
df.head(10)
```

---

## 🟩 Create Remote Column (1.25.2)

```python
# Create a new column is_remote that indicates whether the job is remote (True) or not (False) based on the job_work_from_home column.
df["is_remote"] = df["job_work_from_home"]
df
```

---

## 🟩 Extract Year from Date (1.25.3)

```python
# Create a new column job_posted_year that extracts the year from the job_posted_date column and display the first 5 rows.
df.info() # job_posted_date is object, lets first convert it to datetime
```

```python
df["job_posted_date"] = pd.to_datetime(df["job_posted_date"]) # now job_posted_date is converted to datetime
```

```python
df["job_posted_year"] = df["job_posted_date"].dt.year # dt property and within we can have year property - not a method so dont need (). now it shows the year only; we create new column with month only
```

```python
df.head()
```

---

## 🟨 Drop Missing Salaries (1.25.4)

```python
# Drop rows where the salary_year_avg column has missing values and display the first 5 rows of the cleaned DataFrame.
df.dropna(subset=["salary_year_avg"], inplace=True)
```
