```python
!pip install -U datasets
import pandas as pd
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
```

```python
df
```

---

## 🟩 Calculate Mean Salary (1.26.1)

```python
# Calculate the mean of the salary_year_avg column and display the result.
mean_salary = df["salary_year_avg"].mean()
print(mean_salary)
```

---

## 🟩 Calculate Median Salary (1.26.2)

```python
# Calculate the median of the salary_year_avg column and display the result.
median_salary = df["salary_year_avg"].median()
print(median_salary)
```

---

## 🟨 Average Hourly Salary by Job Title (1.26.3)

```python
# Group the DataFrame by job_title_short and calculate the average salary_hour_avg for each job title. Display the result.
df.groupby("job_title_short")["salary_hour_avg"].mean()
```

---

## 🟨 Job with Lowest Salary (1.26.4)

```python
# Find the job title with the lowest average hourly salary and display the job title and the salary.
lowest_salary_hour_avg = df["salary_hour_avg"].idxmin() # Use the idxmin method to find the index of the minimum value in the salary_hour_avg column
```

```python
df.iloc[lowest_salary_hour_avg] #use iloc to get the corresponding row
```

```python
df.iloc[lowest_salary_hour_avg][["job_title_short", "salary_hour_avg"]]
# The [["job_title_short", "salary_hour_avg"]] segment is applied directly to that extracted row to filter out the unnecessary data. The double brackets are strictly required here because they serve two distinct purposes:

# The outer brackets [ ] represent the standard Pandas syntax for selecting data.

# The inner brackets ["...", "..."] create a standard Python list containing multiple string values.
```

## 🟨 Job Count by Country (1.26.5)

```python
# Group the DataFrame by job_country and calculate the total number of job postings for each country. Display the result.
df.groupby("job_country").size() # Use the size method to count the number of job postings for each group.
```
