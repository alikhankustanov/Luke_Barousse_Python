## 🟨 Job Postings Trend Line Plot (1.31.3)

```python
# Create a line plot showing the trend of job postings over time using the Pandas method to plot:

# Create a DataFrame called monthly_job_counts and use the job_posted_date column to count the number of job postings for each month (in the format 'Jan').
# Add labels for the x-axis (label as 'Month'), y-axis (label as 'Number of Job Postings'), and a title (set as 'Trend of Job Postings Over Time') for the chart.

df["job_posted_month"] = df["job_posted_date"].dt.month
```

```python
df
```

```python
df["job_posted_month"].value_counts()
# df['job_posted_month']: This extracts just the single column containing the month of each job posting. Extracting one column from a DataFrame returns a one-dimensional Pandas Series.

# .value_counts(): This counts how many times each unique month appears in that Series. The output is a new Series where the index (the left side) is the month, and the value (the right side) is the total count of jobs for that month. By default, this method sorts the results from the highest count to the lowest count!
```

```python
monthly_job_counts = df["job_posted_month"].value_counts().sort_index() # create a series
# df['job_posted_month']: This extracts just the single column containing the month of each job posting. Extracting one column from a DataFrame returns a one-dimensional Pandas Series.

# .value_counts(): This counts how many times each unique month appears in that Series. The output is a new Series where the index (the left side) is the month, and the value (the right side) is the total count of jobs for that month. By default, this method sorts the results from the highest count to the lowest count.

# .sort_index(): Because .value_counts() sorts by the highest numbers, your months would be out of chronological order (e.g., November, then March, then July). .sort_index() forces the Series to re-order itself based on the index labels (the months) so that they progress correctly in ascending chronological order.
```

```python
monthly_job_counts
```

```python
monthly_job_counts.plot(kind='line')
plt.ylabel("Number of Job Postings")
plt.xlabel("Months")
plt.title("Trend of Job Postings Over Time")
```
