### Add Items

Since tuples are immutable (not changeable), they don't have a built-in `append()` method. But there are two methods:

**Method 1: Turn tuple into a list.**

* First turn it into a list
* Then use the `append()` method
* Convert the list to a tuple

````python
job_skills = ('python', 'sql', 'statistics', 'tableau')

# Turn tuple into a list
job_skills_list = list(job_skills)

# Add skill to the list
job_skills_list.append('excel')

# Convert list to a tuple
job_skills_tuple = tuple(job_skills_list)

job_skills_tuple
````

**Method 2: Add tuple to a tuple.**

You can add tuples to tuples, which is good if you want to add one or more items.
* Create a new tuple with the items
* Add it to the existing tuple

A for loop in Python is used to iterate over a sequence (such as a list, tuple, dictionary, set, or string) and execute a block of code for each item in that sequence.

**while** loops

A while loop in Python is used to repeatedly execute a block of code as long as a given condition evaluates to True. Unlike a for loop, which runs a fixed number of times over a sequence, a while loop is condition-based and runs until the condition becomes False.

A `while` loop executes a set of statements as long as the condition is true.

Remember to increment `i`, or else the loop will continue forever.

````python
count = 1
while count < 5:
  print(count)
  count += 1
````

````python
# Exercise to train

# Create a function product_sales_summary that takes a list of sales,
# where each sale is a dictionary with keys 'product', 'category', and 'revenue',
# and returns a summary dictionary with the total number of sales, the average revenue,
# and a list of unique categories.

sales_data = [
    {'product': 'Wireless Mouse', 'category': 'Electronics', 'revenue': 25.0},
    {'product': 'Desk Lamp', 'category': 'Home & Office', 'revenue': 45.0},
    {'product': 'Bluetooth Keyboard', 'category': 'Electronics', 'revenue': 60.0},
    {'product': 'Ergonomic Chair', 'category': 'Home & Office', 'revenue': 150.0},
    {'product': 'HDMI Cable', 'category': 'Electronics', 'revenue': 15.0}
]

# Solution:
def product_sales_summary(sales_data):
  total_number_sales = len(sales_data)
  total_revenue = sum(sales["revenue"] for sales in sales_data) # sales is a dictionary in a list. sales["revenue"] is to address the "revenue" key in dictionary
  average_revenue = total_revenue / len(sales_data)
  unique_categories = list(set((sales["category"] for sales in sales_data))) # variable = (list comprehension); i am doing from this the set "unique_categories = (sales["category"] for sales in sales_data)"
  return {
      "total number of sales": total_number_sales,
      "average revenue": average_revenue,
      "unique categories": unique_categories
  }

````

````python
# Create a function library_inventory_summary that takes a list of books.
# Each book is a dictionary with keys 'title', 'author', 'genre', 'pages', and 'is_checked_out'.
#
# The function should return a summary dictionary with:
# 1. "total books": the total number of books in the list.
# 2. "average pages available": the average number of pages of ONLY the books that are currently available
#    (meaning 'is_checked_out' is False).
# 3. "unique genres": a list of the unique values from the 'genre' key across all books.

library_books = [
    {'title': '1984', 'author': 'George Orwell', 'genre': 'Dystopian', 'pages': 328, 'is_checked_out': True},
    {'title': 'The Hobbit', 'author': 'J.R.R. Tolkien', 'genre': 'Fantasy', 'pages': 310, 'is_checked_out': False},
    {'title': 'Dune', 'author': 'Frank Herbert', 'genre': 'Sci-Fi', 'pages': 412, 'is_checked_out': False},
    {'title': 'Brave New World', 'author': 'Aldous Huxley', 'genre': 'Dystopian', 'pages': 268, 'is_checked_out': False},
    {'title': 'Neuromancer', 'author': 'William Gibson', 'genre': 'Sci-Fi', 'pages': 271, 'is_checked_out': True}
]

# Write your function below:
def library_inventory_summary(library_books):
  total_books = len(library_books)
  pages_of_available= sum(books["pages"] for books in library_books if books["is_checked_out"] == False)
  avg_pages_of_available = pages_of_available / sum(1 for books in library_books if books["is_checked_out"] == False)
  unique_genres = list(set((books["genre"] for books in library_books)))
  return {
      "total books": total_books,
      "average pages available": avg_pages_of_available,
      "unique genres": unique_genres
  }
````
