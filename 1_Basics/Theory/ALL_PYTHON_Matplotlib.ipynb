**DataFrame**
A DataFrame is an entire two-dimensional table of data, exactly like an Excel spreadsheet or an SQL database table. It contains multiple rows and columns, and each column can hold a different type of data (such as text for job titles, booleans for remote status, or floats for salaries).

**Series**
A Series is a single, one-dimensional column of data extracted from that table. It is essentially a specialized list where every item shares the exact same data type, accompanied by an index (the row labels).

**The Relationship**
A DataFrame is literally a collection of multiple Series glued together side-by-side.

When you load your entire dataset into df, you are working with a DataFrame.

When you isolate a single column using df["salary_year_avg"], Pandas hands that specific column back to you as a Series.

If you extract a single row using df.iloc[0], Pandas also pivots that row and hands it back to you as a Series, using the original column headers as the new index labels.

# Matplotlib Basics Exercise


```python
# calculate median salary based on the job title
df.groupby("job_title_short")["salary_year_avg"].median()
```

```python
df.groupby("job_title_short")["salary_year_avg"].median().sort_values() # now we have in ascending order from lowest to highest; ,sort_values from lowest to highest by default; .value_counts() from most counts to lowest counts
```

```python
job_salary = df.groupby("job_title_short")["salary_year_avg"].median().sort_values() # job_salary is a series
```

```python
job_salary.plot(kind="barh")
plt.xlabel("Salary($USD)")
plt.ylabel(" ")
plt.title("Median Salary by Job Title")
plt.show()
```
