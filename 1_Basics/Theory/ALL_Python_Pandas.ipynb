# Pandas Basics Exercise

```python
import pandas as pd
from datasets import load_dataset

dataset = load_dataset("lukebarousse/data_jobs")
df = dataset["train"].to_pandas()

df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
```

```python
df
```

```python
df["job_country"].value_counts() # we get the number of mentionings of this countries. so there are 160 rows of diff countries each has different counts
```

```python
df["job_country"].value_counts().head(20) # we list top-20 of countries with highest counts
```

```python
df["job_country"].isin(["Brazil"]) # this iterates over the whole dataframe and output True or False for each row
```

```python
df["job_country"].isin(["Brazil"]).any() # so we use any method to see if at least in 1 row it exists
```

```python
# we wanna create a dataframe with only the country of choice filtered for with that job data
us_jobs = df[df['job_country'] == "United States"] # == comparison operator; df[selection] df["job country"] column job country in dataframe
```

```python
us_jobs
```

```python
# we gonna remove rows in which columns of salary with NaN
us_jobs = us_jobs[us_jobs["salary_year_avg"].notna()]
# If you wrote us_jobs = us_jobs["salary_year_avg"].notna(), you would instantly overwrite your entire DataFrame with a single column of True and False values, losing all your actual job data.

# Filtering in Pandas requires a two-step process known as boolean indexing, which is why your original code nests the condition inside brackets:

# The Inner Test: The segment us_jobs["salary_year_avg"].notna() generates a boolean mask. It evaluates every single row in that specific column and outputs a simple sequence of True (salary exists) or False (salary is NaN).

# The Outer Filter: By wrapping that true/false sequence inside the DataFrame's selection brackets us_jobs[...], you instruct Pandas to apply the mask to the entire table. The DataFrame keeps every complete row that aligns with a True value and drops every row that aligns with a False value.

# To filter a table, you always need both components: the inner mathematical condition to evaluate the data, and the outer DataFrame brackets to execute the physical filtering.
```

```python
us_jobs
```

```python
us_jobs.groupby("job_title_short")["salary_year_avg"].count()
```

```python
us_jobs.groupby("job_title_short")["salary_year_avg"].agg(["min", "max", "median", "count"])
```

```python
# we wanna sort the values by median
us_jobs.groupby("job_title_short")["salary_year_avg"].agg(["min", "max", "median", "count"]).sort_values(by="median", ascending = False)
```
