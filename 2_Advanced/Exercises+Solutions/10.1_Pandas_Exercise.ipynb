```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df["job_skills"] = df["job_skills"]. apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df
```

```python
df_DA = df[df["job_title_short"] == "Data Analyst"].copy() # creating a data frame with job_title_short = Data Analyst; df[filter condition]
```

```python
df_DA
```

```python
df_DA["job_posted_month_no"] = df_DA["job_posted_date"].dt.month # create a new column job_posted_month_no in the df_DA dataframe; 
df_DA
```

```python
df_DA_explode = df_DA.explode("job_skills") # creating a new dataframe with job_skills column exploded
df_DA_explode
```

```python
df_DA_explode.pivot_table(index="job_posted_month_no", columns="job_skills", aggfunc="size") # notice we have NaN values
```

```python
df_DA_explode.pivot_table(index="job_posted_month_no", columns="job_skills", aggfunc="size", fill_value=0) # we fill the NaN values with 0
```

```python
df_DA_pivot = df_DA_explode.pivot_table(index="job_posted_month_no", columns="job_skills", aggfunc="size", fill_value=0)
```

```python
df_DA_pivot.loc["Total"] = df_DA_pivot.sum() # create a new row called Total
df_DA_pivot

# Right side: df_DA_pivot.sum()

# By default, .sum() operates vertically down each column (axis=0).

# It calculates the grand total of all rows for each individual column in df_DA_pivot.



# Left side: df_DA_pivot.loc["Total"] = ...

# .loc["Total"] uses label-based indexing to look for a row indexed as "Total".

# Because a row named "Total" does not exist yet, pandas creates a new row at the bottom of the DataFrame and assigns the column totals to it.
```

```python
df_DA_pivot.loc["Total"].sort_values(ascending=False) # now Total row exists so we can обращаться to it
```

```python
df_DA_pivot.loc["Total"].sort_values(ascending=False).index # to see the indexes
```

```python
df_DA_pivot = df_DA_pivot[df_DA_pivot.loc["Total"].sort_values(ascending=False).index ] # all these columns (above) we have, sort the values by them
df_DA_pivot
```

```python
# we dont need Total row anymore
df_DA_pivot = df_DA_pivot.drop("Total")
df_DA_pivot
```

```python
df_DA_pivot.iloc[:, :5] # iloc[row slicing, column slicing]
```

```python
df_DA_pivot.iloc[:, :5].plot(kind="line")
plt.title("Top 5 Skills for Data Analysts per Month")
plt.xlabel(" ")
plt.ylabel("Count")
plt.show()
```
