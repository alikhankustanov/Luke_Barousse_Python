```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

### Problem #1 Statement:
Create a copy of the DataFrame and drop rows where the salary_year_avg column has missing values. Display the first 5 rows of the cleaned DataFrame.

```python
df_copy = df.copy()
```

```python
df_copy = df_copy.dropna(subset="salary_year_avg") # In that code, the axis is automatically set to 0 by default. This means pandas will drop the entire row if a missing value is found in your specified subset column.
# axis: Use 0 to drop rows with missing values, or 1 to drop columns.
df_copy.head()
```

### Problem #2 Statement:
Create a new DataFrame that is a random sample of 10% of the original DataFrame. Display the first 5 rows of the sampled DataFrame.

```python
df_sampled = df.sample(frac=0.1)
# The .sample() method has a frac parameter. Instead of specifying an exact number of rows with n, frac allows you to extract a random percentage of your entire DataFrame. For example, running df.sample(frac=0.25) will return a random 25% of your dataset. You cannot use n and frac in the same command.
df_sampled.head()
```

### Problem #3 Statement:
Create a copy of the original DataFrame and fill in missing values in the salary_year_avg column with the median salary. Display the first 5 rows of the altered DataFrame.
