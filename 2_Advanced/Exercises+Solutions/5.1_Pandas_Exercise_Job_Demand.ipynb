```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df
```

```python
df_us = df[df["job_country"] == "United States"].copy() # filtering dataset to onyl have USA
# copy so that we create a unique dataframe
# If you do not use .copy(), changes made to df_us might alter the original df, but they also might not. This unpredictability is exactly why pandas throws the SettingWithCopyWarning.

# When you filter data without copying, pandas creates a linked window (a "view") to the original dataset. If you try to modify df_us later—such as changing a salary value or adding a new column—pandas gets confused about your intention. It does not know if you want that modification to bleed through the window and permanently alter the original master df, or if you only meant to change the subset. Depending on how pandas arranged the data in your computer's memory, the original master document might be modified, or only a temporary, invisible copy might be altered and immediately lost.

# Because of this unpredictable behavior, it is never safe to modify a filtered dataset without copying it first. Using .copy() physically duplicates the data into a new, completely disconnected space. This guarantees that no matter what calculations, deletions, or edits you make to the US data, your original master df remains perfectly intact.
```

```python
df_us["job_posted_month"] = df_us["job_posted_date"].dt.strftime("%B") # strftime to have month not like a number but as a stringж to have month we need to use %B
df_us # we added new column job_posted month with month as a string
```

```python
df_us_pivot = df_us.pivot_table(index="job_posted_month", columns="job_title_short", aggfunc="size") # we want to show the number of job postings in each month so use size
# note that index of month is in alphabetical order so we need to keep it in right order
df_us_pivot
```

```python
# we need to get an actual number associated with the this job_posted_month and then sort it by that
df_us_pivot.reset_index(inplace=True)
df_us_pivot
# If your rows are currently labeled by text (which happens automatically when you create a pivot table or group data), this method pushes those text labels back into the DataFrame as a standard column and resets the far-left row index back to the default 0, 1, 2 numerical sequence.
```

```python
df_us_pivot["month_no"] = pd.to_datetime(df_us_pivot["job_posted_month"], format="%B") # we create month no column
# The format parameter tells pandas what the incoming text currently looks like so it knows exactly how to read and translate it into a proper datetime object.
df_us_pivot["month_no"] # we didnt specify the year and day so it just takes 1900 and 01
```

```python
df_us_pivot["month_no"] = pd.to_datetime(df_us_pivot["job_posted_month"], format="%B").dt.month # we only want to extract the month
df_us_pivot["month_no"]
```

```python
df_us_pivot # now it shows for April 4, for August 8 etc.
```

```python
df_us_pivot.sort_values("month_no", inplace=True) # so "month no" column is in ascending order now
df_us_pivot.set_index("job_posted_month", inplace=True)
# now months are in chronological order
df_us_pivot
```

```python
df_us_pivot.drop(columns="month_no", inplace=True)
df_us_pivot
# Here is the exact breakdown of how each component works:

# .drop(): This is the pandas method used specifically to remove data. It can delete either rows or columns depending on the parameters you provide.

# columns="month_no": This tells the function exactly what to remove. By explicitly stating columns, you direct pandas to target the vertical columns rather than the horizontal rows, specifically seeking out the one labeled "month_no".

# inplace=True: This forces pandas to execute the deletion directly inside your existing df_us_pivot variable in memory. If you had omitted this (which defaults to inplace=False), pandas would leave your original DataFrame completely untouched and instead output a temporary copy of the dataset with the column removed.
```

```python
df_us_pivot.plot(kind="line")
```
