```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

### Problem #1 Statement:
Create a DataFrame df_us_jobs that contains only job postings in the United States.
Then, create another DataFrame df_monthly_postings that counts the number of U.S. job postings for each month.
Merge these two DataFrames on the job_posted_month column using a right join, and display the first 5 rows of the resulting DataFrame.

```python
df_us_jobs = df[df["job_country"] == "United States"] # df[selection condition]
df_us_jobs
```

```python
df_us_jobs["job_posted_month"] = df_us_jobs["job_posted_date"].dt.strftime("%B") # added job_posted_month column in the dataframe
df_us_jobs
```

```python
df_monthly_postings = df_us_jobs.groupby("job_posted_month").size()
df_monthly_postings
# Yes, there are two specific issues here that caused the ValueError when you tried to merge:   
# It is a Series, not a DataFrame: Running .size() returns a pandas Series (a single column of numbers), not a 2D DataFrame table.   
# job_posted_month is an Index, not a Column: The month names are sitting in the far-left index position, which means .merge(on="job_posted_month") cannot find a standard column with that name to join on. 
```

```python
df_monthly_postings = df_us_jobs.groupby("job_posted_month").size().reset_index(name="total_postings") #name for the counts
df_monthly_postings
```

```python
df_merged = df_us_jobs.merge(df_monthly_postings, on="job_posted_month")
df_merged.head()
```

### Problem #2 Statement:
Create a DataFrame df_not_remote_jobs that contains only job postings that do not offer remote work (job_work_from_home == False).

From df_not_remote_jobs, calculate the average yearly salary (salary_year_avg) for each job title (job_title_short).
Store the result in a new DataFrame df_not_remote_salary.
Rename the salary column to salary_not_remote_avg.

Merge df_not_remote_jobs and df_not_remote_salary on the job_title_short column using a right join.

Display the first 5 rows of the resulting DataFrame.

```python
df_not_remote_jobs = df[df["job_work_from_home"] == False]
df_not_remote_jobs
```

```python
df_not_remote_salary = df_not_remote_jobs.groupby("job_title_short")["salary_year_avg"].mean()
df_not_remote_salary
# this is not a dataframe
```

```python
df_not_remote_salary = df_not_remote_jobs.groupby("job_title_short")["salary_year_avg"].mean().reset_index()
df_not_remote_salary
```

```python
df_not_remote_salary = df_not_remote_salary.rename(columns={"salary_year_avg":"salary_not_remote_avg"})
df_not_remote_salary

# IMPORTANT: df_not_remote_salary = df_not_remote_salary.rename(columns={"salary_year_avg":"salary_not_remote_avg"}) no need inplace=True because we assigned a variable
# IMPORTANT: df_not_remote_salary.rename(columns={"salary_year_avg":"salary_not_remote_avg"}, inplace=True) inplace is used when no variable assignment and directly inplaces in our dataframe
```

```python
df_merged = df_not_remote_jobs.merge(df_not_remote_salary, on="job_title_short", how="right").head()
# 'right': Keeps all rows from the new df_right, pulling in matching data from the left where available.
df_merged
```

### Problem #3 Statement:
Create a DataFrame df_company_salary that calculates the average yearly salary for each company.
Then, create another DataFrame df_company_remote that counts the number of remote job postings for each company. Name this count column remote_count.
Merge these two DataFrames on the company_name column using an inner join and display the first 5 rows of the resulting DataFrame.

```python
df_company_salary = df.groupby("company_name")["salary_year_avg"].mean().reset_index()
df_company_salary
```

```python
df_company_remote = df[df["job_work_from_home"] == True].groupby("company_name").size().reset_index(name="remote_count") # you groupby not the original df but the df with the remote jobs!
df_company_remote
```

```python
df_merged_final = df_company_salary.merge(df_company_remote, on="company_name", how="inner").head()
# 'inner' (default): Keeps only the rows where the matching key exists in both DataFrames.
df_merged_final
```
