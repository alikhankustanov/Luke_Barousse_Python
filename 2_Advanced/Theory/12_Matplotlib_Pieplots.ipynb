```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt 

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

enumerate() is a built-in Python function that tracks both the index (counter) and the item value simultaneously while looping through a sequence (like a list, tuple, or Series).

Instead of manually creating a counter variable (i = 0) and updating it on each iteration (i += 1), enumerate() automatically yields a tuple of (index, item) for every step.

<br>

job_roles = ['Data Analyst', 'Data Scientist', 'Data Engineer']

for i, role in enumerate(job_roles):
    print(f"Index {i}: {role}")
<br>

will give:
- Index 0: Data Analyst
- Index 1: Data Scientist
- Index 2: Data Engineer

```python
# use subplots to plot all 3 pie charts | USING MATPLOTLIB METHOD

fig, ax = plt.subplots(1, 3) # 1 row 3 columns

# key as the column we want, value as the title we want for each pie plot
dict_column ={
    "job_work_from_home" : "Work from Home",
    "job_no_degree_mention" : "Job Degree Req.",
    "job_health_insurance" : "Health Insurance Offered"
}

for i, (column,title) in enumerate(dict_column.items()):  # enumerate provides as index back; (column, title) is a tuple
     ax[i].pie(df[column].value_counts(), startangle=90, autopct="%1.1f%%", labels=["False", "True"]) # matplotlib syntax for subplots
     ax[i].set_title(title)
 

plt.show()

# for i, (column, title) in enumerate(dict_column.items()):

# dict_column.items(): Extracts (key, value) pairs from the dictionary: ("job_work_from_home", "Work from Home").

# enumerate(...): Adds a zero-based counter (0, 1, 2) to each dictionary pair.

# Unpacking i, (column, title): On each loop iteration, Python automatically breaks down the tuple into three variables:

# i: The numeric index (0 on 1st loop, 1 on 2nd, 2 on 3rd) used to position subplots (ax[i]).

# column: The column key string used to pull data (df[column]).

# title: The chart title string used for set_title().




# ax[i].pie(df[column].value_counts(), startangle=90, autopct="%1.1f%%", labels=["False", "True"])

# ax[i]: Targets the specific subplot box for the current loop (ax[0], ax[1], or ax[2]).

# .pie(...): Matplotlib's method for building pie charts.

# df[column].value_counts(): Provides the numeric slice sizes (counts of False vs True in that column).

# startangle=90: Rotates the first slice so it starts vertically at the 12 o'clock position (90 degrees).

# autopct="%1.1f%%": Automatically calculates percentage values and displays them inside each pie slice with 1 decimal place (e.g., 89.0%).

# labels=["False", "True"]: Assigns category labels to pie slices.
```
