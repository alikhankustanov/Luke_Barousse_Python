```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt 

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df
```

```python
df = df[df["job_title_short"] == "Data Analyst"] # we want to replace the entire dataframe 
```

```python
df_exploded = df.explode("job_skills")

df_exploded
```

```python
df_exploded.groupby("job_skills").size()
```

```python
# we want to show the number of counts of job skills and the median salary so two aggfunctions then

df_exploded.groupby("job_skills").agg(
    skill_count = ("job_skills", "count") #skill count is the column name, job_skills is column we want to do the aggregation on and aggregation method of count
)
```

```python
df_exploded.groupby("job_skills").agg(
    skill_count = ("job_skills", "count"), 
    median_salary = ("salary_year_avg", "median")
)
```

```python
skill_stats = df_exploded.groupby("job_skills").agg(
    skill_count = ("job_skills", "count"), 
    median_salary = ("salary_year_avg", "median")
)

skill_stats
```

```python
skill_stats = skill_stats.sort_values(by="skill_count", ascending=False).head(10)
skill_stats
```

```python
skill_stats.plot(kind="scatter", x="skill_count", y="median_salary")
plt.xlabel("Count of Job Postings")
plt.ylabel("Median Yearly Salary($USD)")
plt.title("Salary vs Count of Job Postings for Top 10 Skills")
plt.tight_layout()
plt.show()

# but how do we see where python, sql, r??
```
