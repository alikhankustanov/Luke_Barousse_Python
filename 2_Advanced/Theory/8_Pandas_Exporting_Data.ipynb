```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df
```

```python
df_DA = df[(df['job_title_short'] == 'Data Analyst')].copy() # df[selection]; df_DA is a df where job_title_short = Data Analyst
df_DA['job_posted_month'] = df_DA['job_posted_date'].dt.strftime('%b') # create a new column job_posted_month 

months = df_DA['job_posted_month'].unique()
df_DA_month = {month: df_DA[df_DA['job_posted_month'] == month].copy() for month in months} # key:value for month in month

quarter_1 = [df_DA_month['Jan'], df_DA_month['Feb'], df_DA_month['Mar']]
df_concat = pd.concat(quarter_1, ignore_index=True)

df_concat
```
