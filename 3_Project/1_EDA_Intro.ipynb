```python
# Importing Libraries
import seaborn as sns
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt 

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df
```

```python
df_DA_US = df[(df["job_country"] == "United States") & (df["job_title_short"] == "Data Analyst")].copy()
```

```python
df_DA_US["job_location"].value_counts().head(10)
```

```python
df_plot = df_DA_US["job_location"].value_counts().head(10).to_frame() # to make it dataframe w/o need of reset_index()!!!
df_plot
```

```python
sns.barplot(data=df_plot, x="count", y="job_location")
```

```python
sns.set_theme(style="ticks")
sns.barplot(data=df_plot, x="count", y="job_location", hue="count", palette="dark:b_r", legend=False) # data is the dataframe itself; or we colud used y=df_plot.index; hue shows waht to color,pallete shows gradient from dark to blue, _r is reverse gradient direction 

# hue (What variable to color by)
# Tells Seaborn which column determines the color of each bar. When you set hue="count", you are asking Seaborn to assign colors dynamically based on the value of the count column.

# palette (What color scheme to apply)
# Tells Seaborn which visual colors/gradient to use for the variable assigned in hue. Without hue, setting palette will either have no effect or trigger a warning in newer Seaborn versions.

sns.despine() # sns.despine(), which removes the top and right border lines to create an even cleaner
plt.title("Counts of Job Locations for Data Analyst in the USA")
plt.xlabel("Number of Jobs")
plt.ylabel(" ")
plt.show()
```

```python
dict_column = {
    'job_work_from_home': 'Work from Home Offered',
    'job_no_degree_mention': 'Degree Requirement',
    'job_health_insurance': 'Health Insurance Offered'
}

fig, ax = plt.subplots(1, 3, figsize=(11, 3.5))

for i, (column, title) in enumerate(dict_column.items()):
    ax[i].pie(df_DA_US[column].value_counts(), labels=['False', 'True'], autopct='%1.1f%%', startangle=90)
    ax[i].set_title(title)

plt.show()
```
